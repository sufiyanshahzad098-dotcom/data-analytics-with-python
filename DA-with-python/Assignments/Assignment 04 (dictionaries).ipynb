
Print keys and values using loop.

**Q4 — Count Frequency of Numbers**

```
numbers = [1,2,2,3,3,3,4,4,4,4]
```

Create frequency dictionary.

````python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print(frequency)
````

**Q5 — Word Counter**

Take sentence input and count word frequency.

````python
sentence = input("Enter a sentence: ")

words = sentence.split()

word_count = {}
for word in words:
    if word in word_count:
        word_count[word] = word_count[word] + 1
    else:
        word_count[word] = 1

print(word_count)
````

**Q6 — Student Marks & Average**

```
marks = {"Math":80, "English":70, "Physics":90}
```

Calculate total and average.

````python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = 0
for subject in marks:
    total = total + marks[subject]

average = total / len(marks)

print("Total marks:", total)
print("Average marks:", average)
````

**Q7 — Highest Scoring Student**

```
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
```

Find highest scoring student.

````python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_name = ""
highest_score = -1

for name in students:
    if students[name] > highest_score:
        highest_score = students[name]
        highest_name = name

print("Highest scoring student:", highest_name, "with", highest_score, "marks")
````

**Q8 — Merge Dictionaries**

```
dict1 = {"a":1,"b":2}
```

Merge and print result.
