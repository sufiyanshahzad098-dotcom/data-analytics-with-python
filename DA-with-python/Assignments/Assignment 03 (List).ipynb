1. Write a program that accepts a list from user and print the alternate element of list.

2. Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.

````python
# Reverse a list without using reverse()
elements = input("Enter list elements separated by space: ").split()

reversed_list = []
for i in range(len(elements) - 1, -1, -1):
    reversed_list.append(elements[i])

print("Reversed list:", reversed_list)
````

3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

````python
numbers_input = input("Enter numbers separated by space: ").split()

numbers = []
for value in numbers_input:
    numbers.append(int(value))

largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

print("Largest number:", largest)
````

4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

````python
elements = input("Enter list elements separated by space: ").split()

last_element = elements[-1]
for i in range(len(elements) - 1, 0, -1):
    elements[i] = elements[i - 1]
elements[0] = last_element

print("Rotated list:", elements)
````

5. Write a program that input a string and ask user to delete a given word from a string.

````python
text = input("Enter a string: ")
word_to_delete = input("Enter the word to delete: ")

words = text.split()
new_words = []
for word in words:
    if word != word_to_delete:
        new_words.append(word)

new_text = " ".join(new_words)
print("Updated string:", new_text)
````

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

````python
date_str = input("Enter date in mm/dd/yyyy format: ")
date_parts = date_str.split("/")
month = date_parts[0]
day = date_parts[1]
year = date_parts[2]

month_names = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

month_name = month_names[int(month) - 1]
formatted_date = month_name + " " + str(int(day)) + ", " + year

print(formatted_date)
````

7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses"

````python
def capitalize_words(sentence):
    words = sentence.split()
    new_words = []
    for word in words:
        new_word = word[0].upper() + word[1:]
        new_words.append(new_word)
    result = " ".join(new_words)
    return result

sentence = input("Enter a sentence: ")
result = capitalize_words(sentence)
print(result)
````

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :

```
2   11   7   12
5    2   9   15
8    3  10   42
```

Sum of row 1 = 32

Sum of row 2 = 31

Sum of row 3 = 63

````python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

matrix = []
for i in range(rows):
    row_input = input("Enter row " + str(i + 1) + " elements separated by space: ").split()
    row = []
    for value in row_input:
        row.append(int(value))
    matrix.append(row)

for i in range(rows):
    row_sum = 0
    for value in matrix[i]:
        row_sum = row_sum + value
    print("Sum of row", i + 1, "=", row_sum)
````

9. Write a program to add two matrices of size n x m.

````python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

print("Enter elements of first matrix:")
matrix1 = []
for i in range(rows):
    row_input = input("Row " + str(i + 1) + ": ").split()
    row = []
    for value in row_input:
        row.append(int(value))
    matrix1.append(row)

print("Enter elements of second matrix:")
matrix2 = []
for i in range(rows):
    row_input = input("Row " + str(i + 1) + ": ").split()
    row = []
    for value in row_input:
        row.append(int(value))
    matrix2.append(row)

result = []
for i in range(rows):
    result_row = []
    for j in range(cols):
        result_row.append(matrix1[i][j] + matrix2[i][j])
    result.append(result_row)

print("Sum of the two matrices:")
for row in result:
    print(row)
````

10. Write a program to multiply two matrices

````python
rows1 = int(input("Enter rows of first matrix: "))
cols1 = int(input("Enter columns of first matrix: "))

print("Enter elements of first matrix:")
matrix1 = []
for i in range(rows1):
    row_input = input("Row " + str(i + 1) + ": ").split()
    row = []
    for value in row_input:
        row.append(int(value))
    matrix1.append(row)

rows2 = int(input("Enter rows of second matrix: "))
cols2 = int(input("Enter columns of second matrix: "))

print("Enter elements of second matrix:")
matrix2 = []
for i in range(rows2):
    row_input = input("Row " + str(i + 1) + ": ").split()
    row = []
    for value in row_input:
        row.append(int(value))
    matrix2.append(row)

if cols1 != rows2:
    print("Matrix multiplication not possible. Columns of first matrix must equal rows of second matrix.")
else:
    result = []
    for i in range(rows1):
        result_row = []
        for j in range(cols2):
            total = 0
            for k in range(cols1):
                total = total + matrix1[i][k] * matrix2[k][j]
            result_row.append(total)
        result.append(result_row)

    print("Product of the two matrices:")
    for row in result:
        print(row)
````
