1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan", "Neha"],
    "Age": [28, 34, 45, 26, 38, 31],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai", "Delhi"],
    "Salary": [45000, 62000, 85000, 38000, 70000, 52000]
}

df = pd.DataFrame(data)

print("First 3 rows:\n", df.head(3))
print("\nShape:", df.shape)
print("\nColumn names:", list(df.columns))
print("\nData types:\n", df.dtypes)
```

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
# working on a copy so the original df stays the same for the next questions
df2 = df.copy()

df2["Bonus"] = df2["Salary"] * 0.10
df2["Total"] = df2["Salary"] + df2["Bonus"]
print("After adding Bonus and Total:\n", df2)

df2 = df2.rename(columns={"City": "Location"})
print("\nAfter renaming City to Location:\n", df2)

df2 = df2.drop(columns=["Bonus"])
print("\nAfter dropping Bonus:\n", df2)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
df_na = pd.DataFrame({
    "Name": ["Amit", "Priya", None, "Sneha", "Karan"],
    "Age": [28, np.nan, 45, 26, 38],
    "Salary": [45000, 62000, np.nan, 38000, np.nan]
})
print("DataFrame with missing values:\n", df_na)

print("\nMissing values in each column:\n", df_na.isnull().sum())

# fill numeric columns with their own mean
num_cols = df_na.select_dtypes(include="number").columns
df_na[num_cols] = df_na[num_cols].fillna(df_na[num_cols].mean())
print("\nAfter filling numeric NaNs with the mean:\n", df_na)

# Name is not numeric so that missing value is still there, drop that row
df_na = df_na.dropna()
print("\nAfter dropping rows that still have missing values:\n", df_na)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
sorted_df = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted by Salary (descending) and then Age (ascending):\n", sorted_df)

print("\nTop 3 highest paid employees:\n", df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
city_group = df.groupby("City")["Salary"]

print("Average salary per city:\n", city_group.mean())
print("\nNumber of employees per city:\n", city_group.count())
print("\nMaximum salary per city:\n", city_group.max())

# all of them together using agg()
result = df.groupby("City")["Salary"].agg(["mean", "count", "max"])
print("\nUsing agg() with multiple functions:\n", result)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3, 4, 5],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan"],
    "DeptID": [10, 20, 10, 30, 40]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 30, 50],
    "DeptName": ["HR", "IT", "Finance", "Sales"]
})

inner = pd.merge(employees, departments, on="DeptID", how="inner")
print("Inner join:\n", inner)

left = pd.merge(employees, departments, on="DeptID", how="left")
print("\nLeft join:\n", left)

# two DataFrames with the same structure (one row is repeated in both)
df_a = pd.DataFrame({"Name": ["Amit", "Priya", "Rahul"], "Age": [28, 34, 45]})
df_b = pd.DataFrame({"Name": ["Rahul", "Sneha", "Karan"], "Age": [45, 26, 38]})

combined = pd.concat([df_a, df_b], ignore_index=True)
print("\nAfter concat:\n", combined)

combined = combined.drop_duplicates()
print("\nAfter removing duplicate rows:\n", combined)
```
